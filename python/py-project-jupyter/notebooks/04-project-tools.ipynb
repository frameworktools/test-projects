## Git hooks and the security audit

`git_hooks_list()` shows the pre-commit checks.

```python
fw.git_hooks_list()
```

This project sits inside the outer `framework-test-projects` repository, so
`git_hooks_install()` refuses: git would run the outer repository's hooks. The error is
expected here.

```python
try:
    fw.git_hooks_install()
except fw.FrameworkError as error:
    print(error)
```

`git_hooks_enable()` refuses the same way and leaves the settings as they were.

```python
try:
    fw.git_hooks_enable("check_notebook_outputs")
except fw.FrameworkError:
    print("refused")
fw.git_hooks_list()[["hook", "enabled"]]
```

`git_security_audit()` checks that private data and secrets stay out of git. It only
reads unless `auto_fix=True`.

```python
fw.git_security_audit(history_depth="shallow").passed
```
